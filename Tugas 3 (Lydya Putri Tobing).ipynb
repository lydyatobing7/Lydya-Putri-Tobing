```python
# List Belanjaan Nobita

list_belanja = ['Apel','Susu','Roti']
harga_jumlah = (5000, 3, 12000, 2, 8000, 1)

print(list_belanja[0])
print(list_belanja[-1])
```

```python
apel = harga_jumlah[0] * harga_jumlah[1]
susu = harga_jumlah[2] * harga_jumlah[3]
roti = harga_jumlah[4] * harga_jumlah[5]
total_belanja = apel + susu + roti
print(total_belanja)
```

```python
nama_pembeli = 'Nobita Nobi'
nama_toko = 'Toko Sembako Doraemon'

print('========================================')
print('         Toko Sembako Doraemon          ')
print('========================================')
print('Pembeli :', nama_pembeli)
print('')
print('Barang \t Harga\tJumlah\tTotal')
print('========================================')
print(f'{list_belanja[0]}\t\t {harga_jumlah[0]}\t  {harga_jumlah[1]}\t{apel}')
print(f'{list_belanja[1]}\t\t {harga_jumlah[2]}\t  {harga_jumlah[3]}\t{susu}')
print(f'{list_belanja[2]}\t\t {harga_jumlah[4]}\t  {harga_jumlah[5]}\t{roti}')
print('----------------------------------------')
print('     Terima Kasih sudah belanja! 🙏     ')
print('========================================')
```
